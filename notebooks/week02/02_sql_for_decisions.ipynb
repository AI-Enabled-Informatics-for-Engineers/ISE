# Week 2 – SQL for Decision Systems
This notebook introduces SQL as a decision-support tool.


```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')
```

```python
df = pd.DataFrame({
    'order_id': [1,2,3,4],
    'region': ['East','West','East','South'],
    'revenue': [100,200,150,50]
})
df.to_sql('orders', conn, index=False)
```

## Decision Question Example
Which region generates the most revenue?

```python
query = '''
SELECT region, SUM(revenue) as total_revenue
FROM orders
GROUP BY region
ORDER BY total_revenue DESC;
'''
pd.read_sql(query, conn)
```
